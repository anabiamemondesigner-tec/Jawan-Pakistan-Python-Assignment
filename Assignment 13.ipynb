# Load Dataset

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("ecommerce_dataset.csv")
df
```

# Check Data Types

```python
df.dtypes
```

# Drop Irrelevant Columns

```python
df.drop(df[["Latitude" , "Longtitude"]] , axis=1  , inplace=True)
```

```python
df
```

# Rename Columns

```python
df.rename(columns={"City Code": "City"}, inplace=True)
df.rename(columns={"Order Priority Code": "Order Priority"}, inplace=True)
df
```

# Duplicate Rows Check

```python
df.duplicated()
```

# Drop Duplicate Rows 

```python
df = df.drop_duplicates()
df
```

# Missing/Null Values Check

```python
df["total"].isnull().sum()
```

### No missing values found in total column

# Missing Sales Value Fill

```python
print("Missing Sales Values:", df["total"].isnull().sum())

median_sales = df["total"].median()

df = df.copy()
df["total"] = df["total"].fillna(median_sales)

print("Missing Sales Values After Filling:", df["total"].isnull().sum())
```

# Top 5 Best Selling Products

```python
top5 = df.groupby("product")["total"].sum().sort_values(ascending=False).head(5)

print(top5)
```

```python
top5 = df.groupby("product")["total"].sum().sort_values(ascending=False).head(5)
top5 = top5.reset_index()

sns.barplot(x="product", y="total", data=top5)

plt.title("Top 5 Products by Sales")
plt.xlabel("Product")
plt.ylabel("Total Sales")
plt.show()
```

##### Headphones has the highest sales among the top 5 products, while Laptop has the lowest sales. Watch are the second highest-selling product.

# Payment Method Distribution

```python
payment_method = df.groupby("payment_method")["total"].sum().sort_values(ascending=False)

print(payment_method)
```

```python
payment_method = (
    df.groupby("payment_method")["total"]
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10, 6))

plt.pie(
    payment_method.values,
    labels=payment_method.index,
    autopct="%1.1f%%",
    shadow=True
)

plt.title("Payment Method Distribution")
plt.axis("equal")
plt.show()
```

#### Credit Card was the most used payment method and generated the highest total sales.

# Month-wise Sales

```python
month = df.groupby("Month")["total"].sum().sort_values(ascending=False)

print(month)
```

```python
month_sales = df.groupby("Month")["total"].sum()

month_order = [
    "January", "February", "March", "April",
    "May", "June", "July", "August",
    "September", "October", "November", "December"
]

month_sales = month_sales.reindex(month_order)

month_sales.plot(kind="line", marker="o")

plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Monthly Sales Trend")
plt.xticks(rotation=45)
plt.show()
```

# City-wise Sales

```python
city_names = {
    1: "Karachi",
    2: "Lahore",
    3: "Islamabad",
    4: "Faisalabad",
    5: "Quetta",
    6: "Peshawar"
}

city_sales = df.groupby("City", dropna=False)["total"].sum().reset_index()
```

```python
sns.barplot(data=city_sales, x="total", y="City")

plt.title("City-wise Sales")
plt.xlabel("Total Sales")
plt.ylabel("City")
plt.show()
print(city_sales)
```

#### Lahore generated the highest sales, while Karachi generated the lowest sales.

# Monthly Sales by Category

```python
monthly_category_sales = (
    df.groupby(["Month", "category"])["total"]
    .sum()
    .reset_index()
)

sns.barplot(
    x="Month",
    y="total",
    hue="category",
    data=monthly_category_sales
)

plt.title("Monthly Sales by Category")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.show()
```
