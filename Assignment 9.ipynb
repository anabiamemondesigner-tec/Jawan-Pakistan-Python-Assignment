Using python csv module add student grades column according to percentages, write data in new csv file including grades column:

Grade
A > 80%–100%

B > 70%–79%

C > 60%–69%

D > 50%–59%

F > below 50%

```python
import csv
```

```python
open_file = open("student data.csv",mode = "r")
open_file
```

```python
reader = csv.reader(open_file)
reader
```

```python
data = list(reader)
data
```

```python
data[0].append("Grade")

for row in data[1:]:
    percentage = float(row[5])

    if percentage >= 80:
        grade = "A"
    elif percentage >= 70:
        grade = "B"
    elif percentage >= 60:
        grade = "C"
    elif percentage >= 50:
        grade = "D"
    else:
        grade = "F"

    row.append(grade)
```

```python
open_write = open("data.csv",mode = "w" , newline = "")
open_write
```

```python
writer = csv.writer(open_write)
writer
```

```python
writer.writerows(data)
```

```python
open_file.close()
open_write.close()
```
