# Q1 : Check Duplicate Records

```python
import pandas as pd
```

```python
df = pd.read_csv("cars.csv")
df
```

```python
duplicate_rows_df = df[df.duplicated()]
print("Numbers of Duplicate Rows:" , duplicate_rows_df.shape)
```

# Q2 : Drop Duplicate Records

```python
df = df.drop_duplicates()
df
```

# Q4 :  Drop missing or null values

```python
df = df.dropna()
df
```

# Q5 : Show top 10 average MSRP (Price) per car brand (rounded figure)?

```python
top_10_brands = df.groupby('Make')['MSRP'].mean().round().sort_values(ascending=False).head(10)

top_10_brands
```

# Q6 :  Show the average MSRP (price) per years (rounded figure)?

```python
df.groupby("Year")["MSRP"].mean().round()
```

# Q7 : 
Show average Engine HP (power) rounded figure and MPG (efficiency) rounded figure per year?

Hint: Average of (city mpg + highway MPG)

```python
df['MPG'] = (df['city mpg'] + df['highway MPG']) / 2

df.groupby('Year').agg({
    'Engine HP': 'mean',
    'MPG': 'mean'
}).round()
```

# Q8 : What percentage of cars are Compact, Midsize, or Large?

```python
df['Vehicle Size'].value_counts(normalize=True).mul(100).round(2)
```

# Q9 : Which vehicle styles are the most sold (most common) in the dataset?

```python
df['Vehicle Style'].value_counts()
```

# Q10 : 
Which car brands have the most vehicles priced at $40,000 or more and with a popularity score of at least 1000?
