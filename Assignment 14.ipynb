```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("database.sqlite")
df = pd.read_sql_query("SELECT * FROM salaries" , conn)
df
```

# 1. 
#### What is the average TotalPayBenefits for each JobTitle? exclude ('Not Provided', 'Not provided')

```python
df = df[
    (df["JobTitle"] != "Not Provided") &
    (df["JobTitle"] != "Not provided")
]

df.groupby("JobTitle")["TotalPayBenefits"].mean()
```

# 2. 
#### Show The JobTitle and Overtime Pay for All Employees With Overtime Pay greater than 50000

```python
df["OvertimePay"] = pd.to_numeric(df["OvertimePay"], errors="coerce")
df[df["OvertimePay"] > 50000][["JobTitle", "OvertimePay"]]
```

# 3.
#### What is the total number of unique employees? exclude ('Not Provided', 'Not provided')

```python
df[
    (df["EmployeeName"] != "Not Provided") &
    (df["EmployeeName"] != "Not provided")
]["EmployeeName"].nunique()
```

## 4. Show The Top 10 highest Paid Employees

```python
df[["EmployeeName" , "TotalPay"]].sort_values(by= "TotalPay" , ascending=(False)).head(10)
```

# 5.
#### Show All Employees Ordered By Their TotalPay Benefits In Descending Order

```python
df[["EmployeeName" , "TotalPayBenefits"]].sort_values(by= "TotalPayBenefits" , ascending=(False))
```
