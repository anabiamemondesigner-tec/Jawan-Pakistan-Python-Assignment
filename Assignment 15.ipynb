```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("factbook.db")
pd.read_sql_query(
    "SELECT name FROM sqlite_master WHERE type='table';",
    conn
)
df = pd.read_sql_query("SELECT * FROM facts", conn)
df
```

# 1. 
### List all cities where the area is greater than 5000, along with their population.

```python
q = '''
SELECT cities.name, cities.population
FROM cities
INNER JOIN facts
ON cities.facts_id = facts.id
WHERE facts.area > 5000
'''

df = pd.read_sql_query(q, conn)
df
```

# 2. 
#### List the names of capital cities where the death rate is greater than 2% and the birth rate is less than 20%

```python
q = '''
SELECT cities.name, facts.death_rate, facts.birth_rate
FROM cities
INNER JOIN facts
ON cities.facts_id = facts.id
WHERE facts.death_rate > 2
AND facts.birth_rate < 20;
'''

df = pd.read_sql_query(q, conn)
df
```

# 3. 
#### List the cities with a population growth rate between 1% and 5%, sorted in descending order by population growth.

```python
q = """
SELECT 
    c.name,
    f.population_growth
FROM cities AS c
INNER JOIN facts AS f
ON c.facts_id = f.id
WHERE f.population_growth BETWEEN 1 AND 5
ORDER BY f.population_growth DESC
"""

df = pd.read_sql_query(q, conn)
df
```

# 4. 
#### Find countries that start with &quot;A&quot; and do not have any cities listed in the cities table. The query should return the country name and city name. Using left join

```python
q = """
SELECT 
    f.name AS country_name,
    c.name AS city_name
FROM facts AS f
LEFT JOIN cities AS c
ON f.id = c.facts_id
WHERE f.name LIKE 'A%'
AND c.name IS NULL
"""

df = pd.read_sql_query(q, conn)
df
```

# 5. 
## List all capital cities and their corresponding death rate using a LEFT JOIN.

```python
q = """
SELECT 
    c.name AS city_name,
    f.death_rate
FROM cities AS c
LEFT JOIN facts AS f
ON c.facts_id = f.id
WHERE c.capital = 1
"""

df = pd.read_sql_query(q, conn)
df
```
