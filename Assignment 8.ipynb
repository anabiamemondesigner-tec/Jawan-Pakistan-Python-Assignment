Make a class called Employee. The __init__() method for Employee should store its different attributes: name, number, address, salary, post, year of joining, education, and his year of experience.

```python
class Employee:
    def __init__(self,name, number, address, salary, post, yearofjoining, education,yearofexperience):
        self.name = name
        self.number = number
        self.address = address
        self.salary = salary
        self.post = post
        self.yearofjoining = yearofjoining
        self.education= education
        self.yearofexperience = yearofexperience

    def describe_employee(self):
        print("Employee Details")
        print(f"Name: {self.name}")
        print(f"Number: {self.number}")
        print(f"Address: {self.address}")
        print(f"Salary: {self.salary}")
        print(f"Post: {self.post}")
        print(f"Year of Joining: {self.yearofjoining}")
        print(f"Education: {self.education}")
        print(f"Year of Experience: {self.yearofexperience}")

    def update_year_of_experience(self):
        self.yearofexperience += self.update_year_of_experience()
        print (f" Updated Year of experience is : {self.yearofexperience} ")

    def increase_salary(self):
        self.salary += self.increase_salary()
        print (f"Your Updated Salary is: {self.salary} ")
```
